The objective of this homework is to solve unconstrained multivariate optimization problems using
line search methods. For this homework, you are going to use the Powell's conjugate direction method.
The specific steps involved are as follows:
1. Starting with coordinate directions, generate a set of conjugate directions using extended parallel
subspace property.
2. Perform exact line search (optimal step size estimation) at each step using Powell's successive
quadratic estimation algorithm.
3. Check for convergence based on change in function value between successive cycles.

```python
import numpy as np
```

```python
def linsearch(f, x, d):
    a, b, c = 0.0, 1.0, 2.0 #a<b<c needed hence this chosen
    fa, fb, fc = f(x+a*d), f(x+b*d), f(x+c*d)
    for _ in range(50): #50 chosen to ensure robustness
        num = (b-a)**2*(fb-fc) - (b-c)**2*(fb-fa)
        den = (b-a)*(fb-fc) - (b-c)*(fb-fa)
        if den == 0: break
        u = b - 0.5*num/den
        fu = f(x+u*d)
        if u < b:
            if fu < fb: c, fc, b, fb = b, fb, u, fu
            else: a, fa = u, fu
        else:
            if fu < fb: a, fa, b, fb = b, fb, u, fu
            else: c, fc = u, fu
    return b

def powell(f, x, tol=1e-6):
    n = len(x)
    D = np.eye(n)
    while True:
        x0, f0 = x.copy(), f(x)
        for i in range(n):
            a = linsearch(f, x, D[i])
            x = x + a*D[i]
        d = x - x0
        a = linsearch(f, x, d)
        x = x + a*d
        D = np.vstack([D[1:], d])
        if abs(f(x) - f0) < tol: break
    return x
```

```python
f1 = lambda x: (x[0]**2 + x[1] - 11)**2 + (x[0] + x[1]**2 - 7)**2
f2 = lambda x: 100*(x[1]-x[0]**2)**2 + (1-x[0])**2
f3 = lambda x: 0.1*((12 + x[0]**2 + (1+x[1])**2)/(x[0]**2 + x[0]**2 + x[1]**2 + 100) + (x[0]*x[1])**4)
f4 = lambda x: (x[0]+10*x[1])**2 + 5*(x[2]-x[3])**2 + (x[1]-2*x[2])**4 + 10*(x[0]-x[3])**4
```

```python
print(powell(f1, np.array([0.0, 0.0]))) #Himmelblau funcn commonly used conditon
print(powell(f2, np.array([-1.0, 1.0]))) #Rosenbroack funcn commonly used condition
print(powell(f3, np.array([1.0, 1.0])))
print(powell(f4, np.zeros(4))) #4 variable polynomial hence initial directions as coordinate axes
```
